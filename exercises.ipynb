```python
## non vectorised max pooling
import math

def avg_pooling(x, k = (3,3), stride = (2,2)):
    # out (1,1,256,256)
    N, C, H_in, W_in = [x.shape[0], x.shape[1], x.shape[2], x.shape[3]]
    kH, kW = k[0], k[1]
    H_out = math.floor((H_in - kH) / stride[0]) + 1
    W_out = math.floor((W_in - kW) / stride[1]) + 1
    out = t.zeros((N, C, H_out, W_out))

    for i in range(N):           # batch
        for j in range(C):       # channel
            for h in range(H_out):
                for w in range(W_out):
                    total = 0
                    for m in range(kH):
                        for n in range(kW):
                            total += input[i, j, stride[0]*h + m, stride[1]*w + n]
                    out[i, j, h, w] = total / (kH * kW)
    return out
    
    
input = t.rand((32, 3, 256,256))
out = avg_pooling(input)
```
